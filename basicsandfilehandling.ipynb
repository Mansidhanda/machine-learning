```python
# Integer
a = 10

# Float
b = 3.14

# String
c = "Hello Mansi"

# List
d = [1, 2, 3, 4]

# Tuple
e = (5, 6, 7)

# Dictionary
f = {"name": "Mansi", "age": 20}

# Boolean
g = True

print(a, b, c, d, e, f, g)
```

```python
print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
print(type(f))
print(type(g))
```

```python
file = open("sample.txt", "w")

file.write("Hello, this is a file handling example.\n")
file.write("We are learning Python.\n")

file.close()

print("Data written to file successfully")
```

```python
file = open("sample.txt", "r")

content = file.read()
print("File Content:\n", content)

file.close()
```

```python
file = open("sample.txt", "a")

file.write("This line is appended.\n")

file.close()

print("Data appended successfully")
```

```python
file = open("sample.txt", "r")

for line in file:
    print(line.strip())

file.close()
```

```python
with open("sample.txt", "r") as file:
    data = file.read()
    print(data)
```

```python
import os

print("File exists:", os.path.exists("sample.txt"))
print("File size:", os.path.getsize("sample.txt"), "bytes")
```
