```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
```

```python
data = {
    'Age': [25,30,35,40,45,50,28,32,38,42],
    'Salary': [20000,25000,30000,35000,40000,45000,22000,27000,32000,37000],
    'Purchased': [0,0,1,1,1,1,0,0,1,1]
}

df = pd.DataFrame(data)
df.to_csv('data.csv', index=False)

print("data.csv file created successfully!")
df
```

```python
df = pd.read_csv('data.csv')
df.head()
```

```python
X = df[['Age', 'Salary']]
y = df['Purchased']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
model = GaussianNB()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print("Predictions:", y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy of Naïve Bayes Classifier:", accuracy)
```

```python
# Example test data
new_data = [[29, 26000]]

prediction = model.predict(new_data)

if prediction[0] == 1:
    print("Customer will Purchase")
else:
    print("Customer will NOT Purchase")
```
