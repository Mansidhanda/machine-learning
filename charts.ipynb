```python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,20,15,25]

plt.plot(x, y)
plt.title("Sales Over Time")
plt.xlabel("Time")
plt.ylabel("Sales")
plt.show()
```

```python
plt.bar(['A','B','C'], [5,7,3])
plt.title("Category Comparison")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
plt.scatter([1,2,3], [4,5,6])
plt.title("Relationship Between Variables")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
```

```python
plt.hist([1,2,2,3,3,3,4])
plt.title("Data Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```
